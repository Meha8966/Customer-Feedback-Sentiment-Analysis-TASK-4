# **Sentiment Analysis Code**

```python
from textblob import TextBlob

# Function to get sentiment
def get_sentiment(text):
    analysis = TextBlob(text)

    if analysis.sentiment.polarity > 0:
        return 'Positive'
    else:
        return 'Negative'

# Apply sentiment analysis
df['Sentiment'] = df['Review'].apply(get_sentiment)

print(df)
```

# **Sentiment Count Graph**

```python
import matplotlib.pyplot as plt

sentiment_count = df['Sentiment'].value_counts()

sentiment_count.plot(kind='bar')

plt.title("Customer Feedback Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Count")

plt.show()
```

# **Insights**
1. Positive reviews are higher than negative reviews.
2. Customer satisfaction can be analyzed using sentiment analysis.
3. Negative feedback helps identify issues.
4. Sentiment analysis improves business decisions.
